```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
print(type(my_list))
my_list.append(6)
print(my_list)
```

```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
print('default:',my_list)
#並且建立常用的實體方法
my_list.append(6)
print('append(6):',my_list)
my_list.remove(2)
print('remove(2):',my_list)
my_list.insert(1, 10)
print('insert(1, 10):',my_list)
my_list.sort()
print('sort',my_list)
my_list.reverse()
print('reverse:',my_list)
my_list.clear()
print('clear:',my_list)
```

```python
#我有一個list,要使用for迴圈,請將list中的每個元素乘以2
my_list = [1, 2, 3, 4, 5]

#for "index" in "這邊放iterable(ex.list、dict) -> 串列資料"
for i in my_list:
    print(i * 2)

for word in "Apple":
    print(word)
```
